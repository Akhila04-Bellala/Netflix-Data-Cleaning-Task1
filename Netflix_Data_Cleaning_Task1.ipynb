```python

# Import the pandas library
import pandas as pd

print("Pandas imported successfully!")
```

```python

# Load the Netflix dataset
df = pd.read_csv('/content/netflix_titles[1].csv')

print("Netflix dataset loaded successfully!")
```

```python

# Display the first 5 rows
df.head()
```

```python

# Check number of rows and columns
print("Dataset Shape:", df.shape)

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

```python

# Display all column names
print(df.columns.tolist())
```

```python

# Understand dataset information
df.info()
```

```python

# Check missing values in each column

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)
```

```python

# Calculate percentage of missing values

missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing Value Percentage:")
print(missing_percentage.round(2))
```

```python

# Create a missing values summary

missing_summary = pd.DataFrame({
    'Missing_Count': df.isnull().sum(),
    'Missing_Percentage': (df.isnull().sum() / len(df)) * 100
})

# Round percentage to 2 decimal places
missing_summary['Missing_Percentage'] = (
    missing_summary['Missing_Percentage'].round(2)
)

# Display summary
missing_summary
```

```python

# Display first 10 records with missing directors

df[df['director'].isnull()].head(10)
```

```python

# Display records with missing cast

df[df['cast'].isnull()].head(10)
```

```python

# Display records with missing country

df[df['country'].isnull()].head(10)
```

```python

# Display records with missing date_added

df[df['date_added'].isnull()]
```

```python

# Display records with missing ratings

df[df['rating'].isnull()]
```

```python

# Display records with missing duration

df[df['duration'].isnull()]
```

```python

# Investigate missing values in important columns

columns_to_check = [
    'director',
    'cast',
    'country',
    'date_added',
    'rating',
    'duration'
]

for column in columns_to_check:
    print("\nMissing records in:", column)
    display(df[df[column].isnull()].head(5))
```

```python

# Display missing record counts and sample titles

for column in ['director', 'cast', 'country',
               'date_added', 'rating', 'duration']:

    missing_rows = df[df[column].isnull()]

    print("\nColumn:", column)
    print("Missing records:", len(missing_rows))
    print("Sample titles:")
    print(missing_rows['title'].head(5).to_string(index=False))
```

```python

# Create a copy of the original dataset

df_clean = df.copy()

print("Clean dataset copy created successfully!")
```

```python

# Replace missing values with Unknown

columns_to_fill = [
    'director',
    'cast',
    'country',
    'rating',
    'duration'
]

for column in columns_to_fill:
    df_clean[column] = df_clean[column].fillna('Unknown')

print("Missing values handled successfully!")
```

```python

# Check missing values after treatment

print("Missing Values After Cleaning:")
print(df_clean.isnull().sum())
```

```python

# Check the number of duplicate rows

duplicate_count = df_clean.duplicated().sum()

print("Total Duplicate Rows:", duplicate_count)
```

```python

# Display duplicate records

df_clean[df_clean.duplicated()]
```

```python

# Remove duplicate rows

df_clean = df_clean.drop_duplicates()

print("Duplicate rows removed successfully!")
```

```python

# Verify duplicates after removal

print("Duplicates After Cleaning:", df_clean.duplicated().sum())

print("Updated Dataset Shape:", df_clean.shape)
```

```python
# Remove extra spaces from column names
df_clean.columns = df_clean.columns.str.strip()

# Convert column names to lowercase
df_clean.columns = df_clean.columns.str.lower()

# Display updated column names
print(df_clean.columns.tolist())
```

```python
df_clean['date_added'] = pd.to_datetime(
    df_clean['date_added'].str.strip(),
    errors='coerce'
)

print(df_clean['date_added'].dtype)
print("Missing dates:", df_clean['date_added'].isnull().sum())
```

```python
df_clean.to_csv('netflix_titles_cleaned.csv', index=False)

print("Cleaned dataset exported successfully!")
```
